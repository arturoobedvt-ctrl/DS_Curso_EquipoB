## 1. Crear un arreglo de NumPy

`np.array()` convierte una lista de Python en un arreglo.

```python
numeros = np.array([1, 2, 3, 4, 5])
print(numeros)
```

## 2. Consultar características del arreglo

`shape` muestra la forma; `ndim`, el número de dimensiones; `size`, la cantidad de elementos.

```python
matriz = np.array([[1, 2, 3], [4, 5, 6]])
print('Matriz:\n', matriz)
print('Forma:', matriz.shape)
print('Dimensiones:', matriz.ndim)
print('Número de elementos:', matriz.size)
```

## 3. Crear arreglos con funciones de NumPy

```python
print('Ceros:\n', np.zeros((2, 3)))
print('Unos:\n', np.ones((2, 3)))
print('Secuencia:', np.arange(0, 10, 2))
```

## 4. Cambiar la forma del arreglo con `reshape()`

```python
valores = np.arange(1, 7)
matriz_2x3 = valores.reshape(2, 3)
print(matriz_2x3)
```
