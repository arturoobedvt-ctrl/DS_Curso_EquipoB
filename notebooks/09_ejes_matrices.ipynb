```python
# Importa NumPy con el alias np para trabajar con arreglos y matrices
import numpy as np

# Dos arreglos de una dimensión, independientes entre sí
m1 = np.array([1, 1])
m2 = np.array([8, 8])

# Muestra cada arreglo por separado
print(m1)
print("")  # línea en blanco para separar los resultados en la consola
print(m2)
print("")

# Une m1 y m2 en una sola secuencia a lo largo del eje 0; resultado: [1 1 8 8]
print(np.concatenate([m1, m2], axis=0))
```
