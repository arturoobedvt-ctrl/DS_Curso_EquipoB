Ejemplo propio: ventas y precios de una tienda de abarrotes. Los conceptos son los del video 4: listas contra arrays.

- **Lista:** acepta datos de distintos tipos y no necesita importar nada.
- **Array:** guarda un solo tipo de dato, necesita un módulo y permite hacer cuentas elemento por elemento.

## 1. Ventas de la semana con el módulo `array`
El primer dato de `array()` es el tipo: `'i'` = enteros, `'d'` = decimales. Sin él, la lista no se acepta.

```python
import array as ar

# Ventas diarias en pesos (lunes a domingo), todas enteras
ventas = ar.array('i', [120, 95, 140, 110, 130, 150, 160])
print(ventas)

# Quitando el tipo, Python marca error: el primer argumento tiene que ser el tipo
try:
    ar.array([120, 95, 140])
except TypeError as error:
    print('Error sin tipo:', error)

# Recorremos el array para imprimir solo los números
for venta in ventas:
    print(venta)
```

## 2. Lo mismo con NumPy
Es la forma que se usa en el curso. No hace falta indicar el tipo y se imprime limpio.

```python
import numpy as np

ventas_np = np.array([120, 95, 140, 110, 130, 150, 160])
print(ventas_np)
```

## 3. Agregar datos: lista contra array
La lista puede guardar texto y números juntos. El array de NumPy no tiene `append`, así que hay que sumar o unir de otra forma.

```python
# Lista de productos: texto y números en la misma lista, eso sí se puede
productos = ['tortilla', 25, 'leche']
productos.append(18)           # agrego un precio más al final
print('Lista:', productos)

precios = np.array([25, 30, 18])

# El array no tiene append; si lo intento, da error
try:
    precios.append(20)
except AttributeError as error:
    print('Error en el array:', error)

# Para subir todos los precios 2 pesos, NumPy aplica la suma a cada elemento
precios = precios + np.array([2])
print('Precios con aumento:', precios)
```

## 4. Operaciones: aquí está la diferencia
Con listas, `+` pega una lista detrás de otra y `*` da error. Con arrays, `+` y `*` se hacen elemento por elemento.

```python
costos = [10, 20, 30]
venta = [15, 25, 35]

# Con listas, + no resta ni suma valores: solo junta las dos listas
print('Lista costos + venta:', costos + venta)

# Multiplicar dos listas no tiene sentido y da error
try:
    print(costos * venta)
except TypeError as error:
    print('Error en lista costos * venta:', error)
```

```python
costos = np.array([10, 20, 30])
venta = np.array([15, 25, 35])

# Ganancia por producto: cada precio de venta menos su costo
print('Ganancia:', venta - costos)

# Piezas vendidas por producto y el ingreso de cada uno (piezas x precio)
piezas = np.array([4, 2, 5])
print('Ingreso:', piezas * venta)
```

## Conclusión
- **Pocos datos o de tipos mezclados** → lista.
- **Muchos datos del mismo tipo y cálculos** → array de NumPy: se opera todo a la vez y ocupa menos memoria.
