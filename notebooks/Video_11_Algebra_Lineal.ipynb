## 1. Matriz y vector columna

```python
import numpy as np

m = np.array([[1, -1, 2], [3, 2, 0]])
m1 = np.array([[1], [2], [3]])
print("Matriz:")
print(m)
print("Vector columna:")
print(m1)
```

**Cómo leer el resultado:** m tiene dos filas y tres columnas. m1 tiene tres filas y una columna; los corchetes dobles permiten representar un vector columna.

## 2. Transposición

```python
# Transponer intercambia las filas y las columnas.
print("Columna transpuesta:", np.transpose(m1))

# En un array de una dimensión, transpose no crea una columna.
m1 = np.array([1, 2, 3])
print("Array de una dimensión transpuesto:", np.transpose(m1))
```

**Cómo leer el resultado:** La columna se convierte en [[1 2 3]], de una fila. El array [1 2 3] sigue siendo de una dimensión después de transponerlo.

## 3. Primer sistema de ecuaciones

```python
# A contiene los coeficientes; b contiene los resultados de las ecuaciones.
A = np.array([[2, 1, -2], [3, 0, 1], [1, 1, -1]])
b = np.array([[-3], [5], [-2]])
print("A:")
print(A)
print("b transpuesto:", np.transpose(b))
x = np.linalg.solve(A, b)
print("Solución:")
print(x)
```

**Cómo leer el resultado:** Cada fila de A es una ecuación. El sistema es 2x + y - 2z = -3; 3x + z = 5; x + y - z = -2. La salida se lee de arriba hacia abajo: x=1, y=-1, z=2.

## 4. Comprobar la solución

```python
# A por x debe recuperar b. allclose tolera pequeños errores decimales.
print("A por x:")
print(np.dot(A, x))
print("¿Coincide con b?:", np.allclose(np.dot(A, x), b))
```

**Cómo leer el resultado:** Se recupera [[-3.], [5.], [-2.]] y aparece True. Eso confirma que la solución satisface las tres ecuaciones.

## 5. Segundo sistema de ecuaciones

```python
m1 = np.array([[2, 7, 3], [2, 8, 2], [1, 3, 1]])
m2 = np.array([1, 1, 0])
# Damos al vector la forma de tres filas y una columna, como en el video.
m2.shape = (3, 1)
print("Matriz:")
print(m1)
print("Resultados:")
print(m2)
solucion = np.linalg.solve(m1, m2)
print("Solución:")
print(solucion)
```

**Cómo leer el resultado:** Se obtiene x=-2, y=0.5, z=0.5. Por ejemplo, en la primera ecuación: 2(-2) + 7(0.5) + 3(0.5) = 1.

## 6. Comprobación adicional

```python
# Comprobación añadida para validar también el segundo ejemplo.
print("¿Coincide con m2?:", np.allclose(np.dot(m1, solucion), m2))
```

**Cómo leer el resultado:** True confirma el segundo sistema. solve requiere una matriz cuadrada no singular; estos dos ejemplos cumplen esa condición.
