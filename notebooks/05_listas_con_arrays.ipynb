Ejemplo propio: temperaturas máximas de tres ciudades mexicanas durante tres días, y un calendario de 15 días.

## 1. Lista a matriz de 2 dimensiones
Una matriz 2D es una lista de filas. Cada corchete interior es una fila: una ciudad, y cada número es un día.

```python
import numpy as np

# Filas: CDMX, Monterrey y Mérida. Columnas: día 1, día 2 y día 3
temperaturas = [[22, 24, 23],
                [35, 36, 33],
                [31, 32, 31]]

# Paso la lista de listas a una matriz de NumPy
m2d = np.array(temperaturas)
print(m2d)
```

## 2. Matriz desde una lista simple
Si la lista tiene un solo nivel, el resultado es una matriz de una dimensión.

```python
# Temperatura máxima de CDMX en cinco días
temp_cdmx = [22, 24, 23, 25, 21]
m1d = np.array(temp_cdmx)
print(m1d)
```

## 3. Forma rápida con `arange` y `reshape`
`arange(15)` crea los números del 0 al 14. Los `reshape(3, 5)` los acomodan en 3 filas (semanas) y 5 columnas (días laborables).

```python
# 15 días de ventas, acomodados en 3 semanas de 5 días
m = np.arange(15).reshape(3, 5)
print(m)
```

Filas por columnas tiene que dar exactamente el número de elementos. Aquí hay 20 elementos y el arreglo es de 3 por 5 (15), así que da error.

```python
try:
    m = np.arange(20).reshape(3, 5)
except ValueError as error:
    print('Error:', error)
```

Con 4 filas por 5 columnas sí cuadra: 4 × 5 = 20. Los números empiezan en 0, por eso el último es el 19.

```python
m = np.arange(20).reshape(4, 5)
print(m)
```

## Conclusión
- Una matriz 2D se escribe como lista de filas.
- Cualquier lista (o lista de listas) se convierte con `np.array(lista)`.
- Con `reshape`, filas × columnas debe ser igual al número de elementos.
