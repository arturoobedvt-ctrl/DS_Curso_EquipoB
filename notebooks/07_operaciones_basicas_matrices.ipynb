```python
import numpy as np
```

## 1. Definir matrices

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])
print('A =\n', A)
print('B =\n', B)
```

## 2. Suma y resta

Estas operaciones se realizan elemento a elemento.

```python
print('A + B =\n', A + B)
print('A - B =\n', A - B)
```

## 3. Multiplicación elemento a elemento y división

```python
print('A * B =\n', A * B)
print('B / A =\n', B / A)
```

## 4. Operaciones con escalares

```python
print('A + 5 =\n', A + 5)
print('A * 2 =\n', A * 2)
```

## 5. Producto matricial

El operador `@` calcula el producto matricial cuando las dimensiones son compatibles.

```python
print('A @ B =\n', A @ B)
```
